# 2.2 Boolean Expressions

A **Boolean expression** evaluates to `true` or `false` — nothing else. The College Board puts it simply: "An expression involving relational operators evaluates to a Boolean value" (College Board, 2025, p. 61).

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `<` | less than |
| `>` | greater than |
| `<=` | less than or equal to |
| `>=` | greater than or equal to |

Run it yourself below.

```java
// CODE_RUNNER: Boolean Expressions - Try It
public class Main {
    public static void main(String[] args) {
        int score = 85;

        System.out.println(score == 85);   // true
        System.out.println(score != 100);  // true
        System.out.println(score > 90);    // false
        System.out.println(score <= 85);   // true
    }
}
Main.main(null);
```

## Raise your hand:
1. If `score` was 90, what would `score > 90` print?
2. Is `score != 85` `true` or `false` right now?

## Where else do you see this?
Every API you build — Spring on the CSA side, any backend on the CSP side — has to decide what to send back before it responds: *Does this exist? Is this value allowed?* Each check is a Boolean expression built from the exact operators above, and the `true`/`false` result decides the response.

```java
if (item == null) {
    return notFound();      // equality operator: does it exist?
}
if (quantity > maxAllowed) {
    return badRequest();    // relational operator: is it valid?
}
```

## AP Exam Practice
Try these on your own before checking the answer.

**1.** What does `12 != 12` evaluate to?
A) `true`  B) `false`  C) `0`  D) Compile-time error

<details><summary>Show answer</summary>B — <code>!=</code> is false when both sides are equal.</details>

**2.**
<pre><code class="language-java">int x = 5, y = 10;
System.out.println(x <= y);</code></pre>
What is printed?
A) `5`  B) `10`  C) `true`  D) `false`

<details><summary>Show answer</summary>C — 5 is less than 10, so the relational expression evaluates to <code>true</code>.</details>

**3.** Which correctly tests two `int` values, `a` and `b`, for equality?
A) `a = b`  B) `a == b`  C) `a.equals(b)`  D) `a != b`

<details><summary>Show answer</summary>B — <code>==</code> compares primitive values directly; <code>a.equals(b)</code> is for objects, not primitives.</details>
