## 1. LxD Cycle Process

**Empathize:** Students send chat messages all day, but the server is a black box: a message just "appears" on other screens. Reviews of novice programming research link difficulty to inaccurate mental models (Qian & Lehman, 2017). Control flow stays abstract when every Unit 2 example is a toy.

**Define:**
- **POV:** CSA students need to see selection and iteration inside software they already use, because a real server makes the `if` and the loop feel necessary instead of decorative.
- **Learning Goal:** Students will trace a message through a chat server, write the checks that guard it, and write the loop that delivers it.

**Ideate:**
- **HMW Question:** How might we make a hidden server step visible as a loop students can run?
- **HMW Question:** How might we show why one open connection beats asking again and again?
- **Activity:** A hands-up poll and a live polling-vs-socket lab, a flowchart walk, runnable Java models of the real server code, and three homework hacks.

**Prototype & Test:** Every Java cell was compiled and run on OpenJDK 21 before publishing. Each homework starter compiles and fails its checks until fixed, and each reference solution passes every check. _Lesson authors: add what happened in your trial run and what you changed._

Here is a model of that guard. Run it, then change one input and predict the result before you run it again.

```java
// CODE_RUNNER: Run the guard on five events. Then change one input and predict the result before you run it again.
public class MessageGuard {
    // A model of the checks in GroupChatWebSocketController
    public static String handle(int groupId, String message, boolean joined) {
        if (groupId < 0) {
            return "ignored: no group";
        }
        if (message == null || message.equals("")) {
            return "ignored: blank message";
        }
        String prefix = "";
        if (!joined) {
            prefix = "auto-join, then ";
        }
        return prefix + "send to /topic/group/" + groupId;
    }

    public static void main(String[] args) {
        System.out.println(handle(7, "hello class", true));
        System.out.println(handle(7, "", true));
        System.out.println(handle(7, null, true));
        System.out.println(handle(-1, "hello class", true));
        System.out.println(handle(7, "hello class", false));
    }
}
MessageGuard.main(null);
```

```java
// CODE_RUNNER: Run the delivery loop. Then change the client count to 5 and predict how many lines print.
public class DeliveryRun {
    public static int broadcast(int clients, String text) {
        int delivered = 0;
        for (int client = 1; client <= clients; client++) {
            System.out.println("client " + client + " received: " + text);
            delivered++;
        }
        return delivered;
    }

    public static void main(String[] args) {
        int count = broadcast(3, "hello class");
        System.out.println("delivered to " + count);
    }
}
DeliveryRun.main(null);
```

#### Hack 3: The Port Bouncer (compound Boolean)

```java
// From ChatWebSocketPortFilter.java (trimmed, reformatted)
boolean isChatWebSocketPath = path.equals(CHAT_ENDPOINT)
        || path.startsWith(CHAT_ENDPOINT + "/");
boolean isOnSocketPort = request.getLocalPort() == socketPort;

// the socket port is only for chat
if (isOnSocketPort && !isChatWebSocketPath) {
    response.setStatus(HttpServletResponse.SC_NOT_FOUND);
    return;
}
// the main port is not for chat
if (!isOnSocketPort && isChatWebSocketPath) {
    response.setStatus(HttpServletResponse.SC_NOT_FOUND);
    return;
}
```
